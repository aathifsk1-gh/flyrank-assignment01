## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision this improves: Which pages should an editor or content lead review and potentially refresh first this week?
Who acts and what they do: A content editor or SEO specialist opens the ranked queue, looks at the top pages, and decides whether to update, expand, or leave them alone.
Cost of a wrong call: If the ranking is bad, the team wastes hours on pages that were fine and misses pages that are quietly losing traffic. Over a few weeks that means lost impressions, wasted writer time, and slower recovery on the pages that actually needed attention.
Why data/ML helps: A simple rule (e.g. “anything down more than 20%”) is easy to write but only moderately accurate. The signals that matter — position, volume, engagement, age, content type — interact in ways that are hard to capture with a few if-statements. A learned score can combine those signals more effectively while still staying transparent enough that an editor can trust (or override) it.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Roughly 54% of the 30,000 pages are already declining. Declining pages still have a solid median impression volume (~961), so they are not just low-traffic noise. About 73% of pages clear a basic volume filter (impressions ≥ 100 and at least one session), which means there is a large, real pool of candidates a ranking system could prioritise.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Dataset:", df.shape[0], "pages,", df.shape[1], "columns")
print()
print("Trend direction breakdown:")
print((df["trend_direction"].value_counts(normalize=True) * 100).round(1).astype(str) + "%")
print()
print("% of pages declining (trend_direction == 'down'):",
      round((df["trend_direction"] == "down").mean() * 100, 1), "%")
print()
print("Median impressions (90d) for declining vs growing pages:")
print(df.groupby("trend_direction")["impressions_90d"].median().loc[["down", "up"]].round(0))
print()
print("% of pages with enough volume to be worth reviewing (impressions >= 100 and sessions > 0):",
      round(((df["impressions_90d"] >= 100) & (df["sessions_90d"] > 0)).mean() * 100, 1), "%")
```

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

Observed patterns in the data (e.g. certain signals appear more often among declining pages).
